### Motivation & Problem Statement

Animal shelters are a community resource, and how well they're funded and staffed tends to track the same neighborhood-level disparities seen in other public services. I want to ask: **does an animal's outcome at an Austin city shelter — time to adoption, or whether it's euthanized — differ depending on the income level of the neighborhood it was found in?**

Practically, if certain neighborhoods consistently produce worse shelter outcomes, that's actionable information for where to target spay/neuter outreach or low-cost vet clinics. Scientifically, it's a human-centered case study in whether resource disparities well-documented elsewhere (health, education) also show up somewhere easy to overlook: public animal welfare. It's the same shape as the course's segregation/COVID-outcomes example — cross-referencing an outcomes dataset against a geographic demographic dataset to check for a disparity. I don't have a fixed hypothesis yet; at this stage my goal is confirming the data can support this kind of comparison and getting a first look at whether any pattern is visible.

### Data Used

**Primary: Austin Animal Center Intakes & Outcomes.** City of Austin's open data portal publishes two linked datasets covering every animal entering the shelter since October 2013: intake/outcome dates and types (adoption, transfer, euthanasia, etc.), animal type/breed/age, and a **"Found Location"** field (address/intersection) for strays.

- Link: data.austintexas.gov ("Austin Animal Center Intakes" / "...Outcomes"); also mirrored on Kaggle under CC0
- License: City of Austin Open Data Terms of Use (intended for public reuse); Kaggle mirror is explicitly CC0. I'll confirm exact City terms wording before hosting data in the repo.

**Secondary: U.S. Census ACS**, 5-year median household income by ZIP Code Tabulation Area (table B19013), via data.census.gov. A U.S. government work — public domain, no restrictions.

**Why this fits:** answering the question requires one dataset of geographically-located animal outcomes and one of neighborhood income to join against it.

**Ethical considerations:** the shelter data has no owner/personal information, keeping privacy risk low. The bigger risk is interpretive — I need to avoid framing results as "this neighborhood doesn't take care of its animals," which would stigmatize under-resourced communities. I'll frame findings around service access and infrastructure, not residents' choices. I'll also watch for the ecological fallacy: ZIP-level income isn't any individual owner's income, so conclusions stay at the neighborhood level.

### Unknowns & Dependencies

1. **Geocoding "Found Location."** Free text, not a clean ZIP code — needs parsing/geocoding to join with Census data; some records won't geocode cleanly. Fallback: aggregate by City Council district if ZIP-level geocoding proves unreliable.
2. **License confirmation.** Confirm exact City of Austin terms before publicly re-hosting data in the repo.
3. **Confounding variables.** Breed mix, distance to shelter, intake volume could all explain an income correlation — named as limitations rather than overclaimed as causation.
4. **Scope.** 100,000+ records since 2013; may need to limit date range to keep geocoding tractable.
